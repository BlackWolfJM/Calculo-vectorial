# Elipse: teoría, elementos y visualización

Este notebook complementa el archivo teórico del subtema. Permite calcular los elementos de una elipse y comprobar su gráfica con Python.

## Objetivos

- Distinguir una elipse horizontal de una vertical.
- Calcular centro, vértices, co-vértices, focos y excentricidad.
- Representar la curva sin duplicar código.


## Recordatorio teórico

Para una elipse con centro $(h,k)$ y $a>b>0$:

- Horizontal: $\frac{(x-h)^2}{a^2}+\frac{(y-k)^2}{b^2}=1$.
- Vertical: $\frac{(x-h)^2}{b^2}+\frac{(y-k)^2}{a^2}=1$.
- Relación focal: $c^2=a^2-b^2$.
- Excentricidad: $e=c/a$, con $0<e<1$.


```python
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt

@dataclass
class Elipse:
    a: float
    b: float
    h: float = 0
    k: float = 0
    eje: str = "horizontal"

    def __post_init__(self):
        if self.a <= 0 or self.b <= 0:
            raise ValueError("a y b deben ser positivos")
        if self.a < self.b:
            raise ValueError("Debe cumplirse a >= b")
        if self.eje not in {"horizontal", "vertical"}:
            raise ValueError("eje debe ser 'horizontal' o 'vertical'")

    @property
    def c(self):
        return np.sqrt(self.a**2 - self.b**2)

    @property
    def excentricidad(self):
        return self.c / self.a

    def elementos(self):
        if self.eje == "horizontal":
            vertices = [(self.h - self.a, self.k), (self.h + self.a, self.k)]
            covertices = [(self.h, self.k - self.b), (self.h, self.k + self.b)]
            focos = [(self.h - self.c, self.k), (self.h + self.c, self.k)]
        else:
            vertices = [(self.h, self.k - self.a), (self.h, self.k + self.a)]
            covertices = [(self.h - self.b, self.k), (self.h + self.b, self.k)]
            focos = [(self.h, self.k - self.c), (self.h, self.k + self.c)]

        return {
            "centro": (self.h, self.k),
            "vertices": vertices,
            "covertices": covertices,
            "focos": focos,
            "excentricidad": self.excentricidad,
        }

    def puntos(self, n=600):
        theta = np.linspace(0, 2*np.pi, n)
        if self.eje == "horizontal":
            x = self.h + self.a*np.cos(theta)
            y = self.k + self.b*np.sin(theta)
        else:
            x = self.h + self.b*np.cos(theta)
            y = self.k + self.a*np.sin(theta)
        return x, y

    def graficar(self):
        x, y = self.puntos()
        datos = self.elementos()
        vx, vy = zip(*datos["vertices"])
        cx, cy = zip(*datos["covertices"])
        fx, fy = zip(*datos["focos"])

        fig, ax = plt.subplots(figsize=(7, 7))
        ax.plot(x, y, color="royalblue", linewidth=2, label="Elipse")
        ax.scatter(*datos["centro"], color="black", marker="x", s=80, label="Centro")
        ax.scatter(vx, vy, color="crimson", label="Vértices")
        ax.scatter(cx, cy, color="darkorange", label="Co-vértices")
        ax.scatter(fx, fy, color="seagreen", label="Focos")
        ax.axhline(0, color="gray", linewidth=0.8)
        ax.axvline(0, color="gray", linewidth=0.8)
        ax.set_aspect("equal", adjustable="box")
        ax.grid(alpha=0.3)
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        ax.set_title(f"Elipse {self.eje}: centro=({self.h}, {self.k})")
        ax.legend()
        plt.show()

        return datos
```

## Ejemplo 1 — Elipse horizontal

Analiza y grafica una elipse con $a=5$, $b=3$ y centro en el origen.


```python
elipse_horizontal = Elipse(a=5, b=3, h=0, k=0, eje="horizontal")
datos_horizontal = elipse_horizontal.graficar()
datos_horizontal
```

## Ejemplo 2 — Elipse vertical desplazada

Ahora el centro es $(2,1)$ y el eje mayor es vertical.


```python
elipse_vertical = Elipse(a=5, b=3, h=2, k=1, eje="vertical")
datos_vertical = elipse_vertical.graficar()
datos_vertical
```
