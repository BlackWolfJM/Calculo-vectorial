## **HIPÉRBOLA**

## 🔹 Forma horizontal (eje mayor en x)

$$
\frac{(x - h) ^ 2}{a^2} -
 \frac{(y - k)^2}{b^2} = 1
$$

- Focos:
$$
(h \pm c, k)
$$

- Relación:
$$
c^2 = a^2 + b^2
$$

```python
import numpy as np
import matplotlib.pyplot as plt

class  Hiperbola:

#insertar datos
    def __init__(self, a, b, h, k):
        self.a = a
        self.b = b
        self.h = h
        self.k = k
#calcular centros
    def centro(self):
      return (self.h , self.k)

#calcular distancia entre los focos
    def calcular_c(self):
      self.c = ( (self.a ** 2) + (self.b ** 2) )** 0.5

#calcular excentrecidad
    def calcular_e(self):
      self.e = self.c /  self.a

#calcular focos
    def calcular_focos(self):
      self.foco1 = (self.h + self.c, self.k)
      self.foco2 = (self.h - self.c, self.k)
      return (self.foco1, self.foco2)

    def graficar_hiperbola(self):
      t = np.linspace(-2, 2, 200)

    # rama derecha
      x1 = self.h + self.a * np.cosh(t)
      y1 = self.k + self.b * np.sinh(t)

    # rama izquierda
      x2 = self.h - self.a * np.cosh(t)
      y2 = self.k + self.b * np.sinh(t)

      plt.figure()
      plt.plot(x1, y1)
      plt.plot(x2, y2)

      plt.scatter(self.h, self.k, color='red')
      plt.gca().set_aspect('equal')
      plt.grid()
      plt.title("Hiperbola")
      plt.show()

    def mostrar_datos(self):

      self.calcular_c()
      self.calcular_e()
      foco1, foco2 = self.calcular_focos()

      print("=" * 50)
      print("        DATOS DE LA HIPERBOLA")
      print("=" * 50)
      print(f"Centro:           ({self.h:.2f}, {self.k:.2f})")
      print(f"Semi-eje real:    a = {self.a:.2f}")
      print(f"Semi-eje imag.:   b = {self.b:.2f}")
      print(f"Distancia focal:  c = {self.c:.2f}")
      print(f"Excentricidad:    e = {self.e:.2f}")
      print("-" * 50)
      print(f"Foco 1:           ({foco1[0]:.2f}, {foco1[1]:.2f})")
      print(f"Foco 2:           ({foco2[0]:.2f}, {foco2[1]:.2f})")
      print("-" * 50)

      print("Asíntotas:")
      print(f"y - {self.k} = ( {self.b/self.a:.2f} )(x - {self.h})")
      print(f"y - {self.k} = -( {self.b/self.a:.2f} )(x - {self.h})")

      print("=" * 50)

      self.graficar_hiperbola()

```

```python
hiperbola1 = Hiperbola(4, 3, 1, -2)

hiperbola1.mostrar_datos()

```

## 🔹 Forma horizontal (eje mayor en x)

$$
\frac{(y - k) ^ 2}{a^2} -
 \frac{(x - h)^2}{b^2} = 1
$$

- Focos:
$$
(h , k \pm c)
$$

- Relación:
$$
c^2 = a^2 + b^2
$$

```python
import numpy as np
import matplotlib.pyplot as plt

class  Hiperbola:

#insertar datos
    def __init__(self, a, b, h, k):
        self.a = a
        self.b = b
        self.h = h
        self.k = k
#calcular centros
    def centro(self):
      return (self.h , self.k)

#calcular distancia entre los focos
    def calcular_c(self):
      self.c = ( (self.a ** 2) + (self.b ** 2) )** 0.5

#calcular excentrecidad
    def calcular_e(self):
      self.e = self.c /  self.a

#calcular focos
    def calcular_focos(self):
      self.foco1 = (self.h, self.k + self.c)
      self.foco2 = (self.h, self.k - self.c)
      return (self.foco1, self.foco2)

    def graficar_hiperbola(self):
      t = np.linspace(-3, 3, 200)

    # rama derecha
      x1 = self.h + self.b * np.sinh(t)
      y1 = self.k + self.a * np.cosh(t)

    # rama izquierda
      x2 = self.h + self.b * np.sinh(t)
      y2 = self.k - self.a * np.cosh(t)

      plt.figure()
      plt.plot(x1, y1)
      plt.plot(x2, y2)

      plt.scatter(self.h, self.k, color='red')
      plt.gca().set_aspect('equal')
      plt.grid()
      plt.title("Hiperbola")
      plt.show()

    def mostrar_datos(self):

      self.calcular_c()
      self.calcular_e()
      foco1, foco2 = self.calcular_focos()

      print("=" * 50)
      print("        DATOS DE LA HIPERBOLA")
      print("=" * 50)
      print(f"Centro:           ({self.h:.2f}, {self.k:.2f})")
      print(f"Semi-eje real:    a = {self.a:.2f}")
      print(f"Semi-eje imag.:   b = {self.b:.2f}")
      print(f"Distancia focal:  c = {self.c:.2f}")
      print(f"Excentricidad:    e = {self.e:.2f}")
      print("-" * 50)
      print(f"Foco 1:           ({foco1[0]:.2f}, {foco1[1]:.2f})")
      print(f"Foco 2:           ({foco2[0]:.2f}, {foco2[1]:.2f})")
      print("-" * 50)

      print("Asíntotas:")
      print(f"y - {self.k} = ( {self.b/self.a:.2f} )(x - {self.h})")
      print(f"y - {self.k} = -( {self.b/self.a:.2f} )(x - {self.h})")

      print("=" * 50)

      self.graficar_hiperbola()

```

```python
hiperbola2 = Hiperbola(4, 3, 1, -2)

hiperbola2.mostrar_datos()
```
