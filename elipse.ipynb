
# **ELIPSE**



## 🔹 Forma horizontal (eje mayor en x)

$$
\frac{(x - h) ^ 2}{a^2} + \frac{(y - k)^2}{b^2} = 1
$$

- Focos:
$$
(h \pm c, k)
$$

- Relación:
$$
c^2 = a^2 - b^2
$$

**insertarndo formulas en las clases**

```python
import numpy as np
import matplotlib.pyplot as plt

class Elipse:
#insertar datos
    def __init__(self, a, b, h, k):
        self.a = a
        self.b = b
        self.h = h
        self.k = k
#calcular el centro
    def centro(self):
        return (self.h, self.k)
#calcular la distancia focal o c
    def calcular_c(self):
        self.c = (self.a ** 2 - self.b ** 2) ** 0.5
#calcular la excentricidad
    def calcular_e(self):
        self.e = self.c / self.a

#calcular cordenadas de focos
    def calcular_focos(self):
        self.foco1 = (self.h + self.c, self.k)
        self.foco2 = (self.h - self.c, self.k)
        return self.foco1, self.foco2
#calcular longitud del eje mayor
    def longitud_eje_mayor(self):
        return 2 * self.a

#calcular longitud de eje menor
    def longitud_eje_menor(self):
        return 2 * self.b

#graficar elipse
    def graficar_elipse(self):
        t = np.linspace(0, 2 * np.pi, 200)
        x = self.h + self.a * np.cos(t)
        y = self.k + self.b * np.sin(t)
        plt.figure()
        plt.plot(x, y)
        plt.scatter(self.h, self.k, color='red', zorder=5)
        plt.gca().set_aspect('equal')
        plt.grid()
        plt.title("Elipse")
        plt.show()

#mostrar datos
    def mostrar_datos(self):

        self.calcular_c()
        self.calcular_e()
        foco1, foco2 = self.calcular_focos()

        print("=" * 50)
        print("           DATOS DE LA ELIPSE")
        print("=" * 50)
        print(f"Centro:           ({self.h:.2f}, {self.k:.2f})")
        print(f"Semi-eje mayor:   a = {self.a:.2f}")
        print(f"Semi-eje menor:   b = {self.b:.2f}")
        print(f"Distancia focal:  c = {self.c:.2f}")
        print(f"Excentricidad:    e = {self.e:.2f}")
        print("-" * 50)
        print(f"Foco 1:           ({foco1[0]:.2f}, {foco1[1]:.2f})")
        print(f"Foco 2:           ({foco2[0]:.2f}, {foco2[1]:.2f})")
        print("-" * 50)
        print(f"Eje mayor:        {self.longitud_eje_mayor():.2f}")
        print(f"Eje menor:        {self.longitud_eje_menor():.2f}")
        print("=" * 50)

        self.graficar_elipse()


```

**creando objetos**

```python
elipse1 = Elipse(a=5, b=3, h=0, k=0)

elipse1.mostrar_datos()
```

```python
elipse2 = Elipse(a= 25 , b= 16, h=9, k=12)

elipse2.mostrar_datos()
```

## 🔹 Forma vertical (eje mayor en y)

$$
\frac{(x - h) ^ 2}{b^2} + \frac{(y - k)^2}{a^2} = 1
$$

- Focos:
$$
(h , k \pm c )
$$

- Relación:
$$
c^2 = a^2 - b^2
$$

**insertarndo formulas en las clases**

```python
import numpy as np
import matplotlib.pyplot as plt

class Elipse:
#insertar datos
    def __init__(self, a, b, h, k):
        self.a = a
        self.b = b
        self.h = h
        self.k = k
#calcular el centro
    def centro(self):
        return (self.h, self.k)
#calcular la distancia focal o c
    def calcular_c(self):
        self.c = (self.a ** 2 - self.b ** 2) ** 0.5
#calcular la excentricidad
    def calcular_e(self):
        self.e = self.c / self.a

#calcular cordenadas de focos
    def calcular_focos(self):
        self.foco1 = (self.h , self.k +  self.c)
        self.foco2 = (self.h , self.k - self.c)
        return self.foco1, self.foco2
#calcular longitud del eje mayor
    def longitud_eje_mayor(self):
        return 2 * self.a

#calcular longitud de eje menor
    def longitud_eje_menor(self):
        return 2 * self.b

#graficar elipse
    def graficar_elipse(self):
        t = np.linspace(0, 2 * np.pi, 200)
        y = self.k + self.a * np.cos(t)
        x = self.h + self.b * np.sin(t)
        plt.figure()
        plt.plot(x, y)
        plt.scatter(self.h, self.k, color='red', zorder=5)
        plt.gca().set_aspect('equal')
        plt.grid()
        plt.title("Elipse")
        plt.show()

#mostrar datos
    def mostrar_datos(self):

        self.calcular_c()
        self.calcular_e()
        foco1, foco2 = self.calcular_focos()

        print("=" * 50)
        print("           DATOS DE LA ELIPSE")
        print("=" * 50)
        print(f"Centro:           ({self.h:.2f}, {self.k:.2f})")
        print(f"Semi-eje mayor:   a = {self.a:.2f}")
        print(f"Semi-eje menor:   b = {self.b:.2f}")
        print(f"Distancia focal:  c = {self.c:.2f}")
        print(f"Excentricidad:    e = {self.e:.2f}")
        print("-" * 50)
        print(f"Foco 1:           ({foco1[0]:.2f}, {foco1[1]:.2f})")
        print(f"Foco 2:           ({foco2[0]:.2f}, {foco2[1]:.2f})")
        print("-" * 50)
        print(f"Eje mayor:        {self.longitud_eje_mayor():.2f}")
        print(f"Eje menor:        {self.longitud_eje_menor():.2f}")
        print("=" * 50)

        self.graficar_elipse()
```
